```python
from tensorflow.keras.models import load_model

image_model = load_model("/content/drive/MyDrive/Image/CDD.h5")
```

```python
from tensorflow.keras.preprocessing import image
import numpy as np

img_path = "/content/drive/MyDrive/content/Comprehensive Disaster Dataset(CDD)/Damaged_Infrastructure/Earthquake/05_02_0001.png"  # change to your image path
img = image.load_img(img_path, target_size=(224,224))  # match your model's input size
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0) / 255.0
```

```python
pred = image_model.predict(img_array)
pred_class = np.argmax(pred)  # if it’s categorical
print("Predicted class index:", pred_class)
```
